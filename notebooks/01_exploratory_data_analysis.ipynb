```python
import pandas as pd

df = pd.read_excel(r"C:\RetailData\online_retail_II.xlsx")

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())
```

```python
missing = df.isnull().sum()

missing_table = pd.DataFrame({
    "Missing Values": missing,
    "Missing %": (missing / len(df) * 100).round(2)
})

missing_table.sort_values(
    by="Missing Values",
    ascending=False
)
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
df[["Quantity", "Price", "Revenue"]].head(10)
```

```python
total_revenue = df["Revenue"].sum()

print(f"Total transaction revenue: £{total_revenue:,.2f}")
```

```python
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")
```

```python
monthly_sales = (
    df.groupby("YearMonth")["Revenue"]
      .sum()
      .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales["YearMonth"] = monthly_sales["YearMonth"].astype(str)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["YearMonth"],
    monthly_sales["Revenue"]
)

plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.title("Monthly Revenue Trend")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
country_sales = (
    df.groupby("Country")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

country_sales
```

```python
plt.figure(figsize=(10, 6))

country_sales.sort_values().plot(kind="barh")

plt.xlabel("Revenue (£)")
plt.ylabel("Country")
plt.title("Top 10 Countries by Revenue")

plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Description")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

```python
plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.xlabel("Revenue (£)")
plt.ylabel("Product")
plt.title("Top 10 Products by Revenue")

plt.tight_layout()
plt.show()
```
